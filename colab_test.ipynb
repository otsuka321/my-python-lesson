```python
# ==================================================
# 【テスト教材】Pythonの「変数（へんすう）」を学ぼう！
# ==================================================
# 変数とは、データ（値や文字）を入れておく「箱」のようなものです。
#
# 下のコードでは、`apple_price` という箱に「150」という数字を入れ、
# `count` という箱に「3」という数字を入れています。
# 自由に数字を書き換えて、左側の「再生ボタン（実行）」を押してみましょう！

# 1. 変数に値を代入する
apple_price = 150
count = 3

# 2. 変数を使って計算する
total_price = apple_price * count

# 3. 結果を画面に表示する
print("リンゴの1個の値段:", apple_price, "円")
print("買った個数:", count, "個")
print("合計金額は:", total_price, "円です！")
```
